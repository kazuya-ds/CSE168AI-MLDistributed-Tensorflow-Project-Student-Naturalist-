```python
# Import libraries
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2

from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
```

```python

# Define path to your images folder
image_path = 'path_to_your_image.jpg'   # change this
img = load_img(image_path) 
x = img_to_array(img)
x = x.reshape((1,) + x.shape)  # reshape for ImageDataGenerator

```

```python
# Define augmentation generator
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,      # If appropriate for your domain
    brightness_range=(0.5, 1.5),
    fill_mode='nearest'
)
```

```python
#Show Augmentaed images# Show augmented images
i = 0
for batch in datagen.flow(x, batch_size=1, 
                          save_to_dir='preview', # directory to save augmented images
                          save_prefix='aug', 
                          save_format='jpeg'):
    i += 1
    if i >= 5:
        break

# Display one example
plt.imshow(batch[0])
plt.axis('off')
plt.title('Augmented Example')
plt.show()
```
